```python
# psed_vanilla
psed_vanilla_s = [12.446736574172974, 118.97893047332764, 412.59264969825745, 745.3477964401245] # *!
psed_vanilla_iter = [30741150, 274334823, 914233016, 1565817691] # *!

# psed_parallel_bsf
psed_parallel_bsf_s = [0.6754541397094727, 1.5370829105377197, 4.026179790496826, 4.605412006378174] # *!
psed_parallel_bsf_iter = [2992455, 20269442, 56320829, 94576636] # *!

# psed_parallel_bsf_lb
psed_parallel_bsf_lb_s = [1.3535232543945312, 3.3911051750183105, 8.266021490097046, 10.376680135726929] # *!
psed_parallel_bsf_lb_iter = [2797205, 19110543, 53303752, 90047349] # *!

# psdtw_vanilla
psdtw_vanilla_s = [419.21485781669617, 4559.293148994446, 15245.766177892685, 30396.453424215317] # *!
psdtw_vanilla_iter = [32540659, 274994784, 880764560, 1513184734] # *!

# psdtw_parallel_bsf
psdtw_parallel_bsf_s = [8.982324600219727, 44.92573142051697, 99.09320592880249, 186.47276854515076] # *!
psdtw_parallel_bsf_iter = [4569737, 30152958, 74702661, 128851112] # *!

# psdtw_parallel_bsf_lb
psdtw_parallel_bsf_lb_s = [9.242284059524536, 43.662862062454224, 92.33867287635803, 182.72490882873535] # *!
psdtw_parallel_bsf_lb_iter = [4156260, 28058604, 70500069, 123040644] # *!
```

```python
for i in range(4):
    time_for_lb_psed = psed_parallel_bsf_lb_s[i] - (psed_parallel_bsf_s[i] / psed_parallel_bsf_iter[i]) * psed_parallel_bsf_lb_iter[i]
    print(
        f"Time for LB (PSED): {time_for_lb_psed:.3f} "
        f"Time proportion for LB (PSED): {time_for_lb_psed/psed_parallel_bsf_lb_s[i]:.3f} "
    )
print("\n")
for i in range(4):
    time_for_lb_psdtw = psdtw_parallel_bsf_lb_s[i] - (psdtw_parallel_bsf_s[i] / psdtw_parallel_bsf_iter[i]) * psdtw_parallel_bsf_lb_iter[i]
    print(
        f"Time for LB (PSDTW): {time_for_lb_psdtw:.3f} "
        f"Time proportion for LB (PSDTW): {time_for_lb_psdtw/psdtw_parallel_bsf_lb_s[i]:.3f} "
    )
```

```python
fig, ax = plt.subplots()

# === Color & Style Definitions ===
# Colors distinguish the specific implementation/variant
c_vanilla = "tab:blue"
c_parallel = "tab:red"
c_parallel_lb = "tab:green"

# === Plotting PSED (Solid Lines) ===
ax.plot(x, psed_vanilla_s, color=c_vanilla, marker="o", linestyle="-", label=r"PSED_vanilla")
ax.plot(x, psed_parallel_bsf_s, color=c_parallel, marker="^", linestyle="-", label=r"PSED_parallel_bsf")
ax.plot(x, psed_parallel_bsf_lb_s, color=c_parallel_lb, marker="s", linestyle="-", label=r"PSED_parallel_bsf_lb")

# === Plotting PSDTW (Dashed Lines) ===
ax.plot(x, psdtw_vanilla_s, color=c_vanilla, marker="o", linestyle="--", label=r"PSDTW_vanilla")
ax.plot(x, psdtw_parallel_bsf_s, color=c_parallel, marker="^", linestyle="--", label=r"PSDTW_parallel_bsf")
ax.plot(x, psdtw_parallel_bsf_lb_s, color=c_parallel_lb, marker="s", linestyle="--", label=r"PSDTW_parallel_bsf_lb")

ax.set_xlabel("Scaling Factor $l$")
ax.set_ylabel("Runtime (s)")

# ax.set_yscale("log")

ax.set_xticks([1.25, 1.50, 1.75, 2.0])

ax.grid(True, which="both", ls="--", lw=0.5, alpha=0.6)

# === Legend ===
# Placed outside to prevent covering data if the plot is crowded
ax.legend(loc="upper left")

plt.tight_layout()
# plt.savefig("../figures/gunpoint-runtime-P2-time.pdf", format="pdf", bbox_inches="tight")
plt.show()
```
